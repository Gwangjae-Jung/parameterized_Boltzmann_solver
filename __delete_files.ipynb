```python
import  os

def remove_files(path: str) -> None:
    for file in sorted(os.listdir(path)):
        name, ext = file.split('.')
        if ext=='ipynb':    continue
        index = int(name[-3:])
        if index%100!=0 or name[:10]=="state_dict":
            os.remove(os.path.join(path, file))
    return None

for folder in os.listdir():
    if os.path.isdir(folder) and folder.startswith('checkpoint__'):
        print(folder)
        remove_files(folder)
```
