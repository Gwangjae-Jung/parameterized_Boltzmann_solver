```python
import  numpy   as  np
import  matplotlib.pyplot  as  plt
```

```python
data_1 = np.ones((10, 10)) + np.random.randn(10, 10) * 0.1
data_2 = 2*np.ones((10, 10)) + np.random.randn(10, 10)

mean_1 = np.mean(data_1, axis=0)
std_1  = np.std(data_1, axis=0)
mean_2 = np.mean(data_2, axis=0)
std_2  = np.std(data_2, axis=0)
```

```python
x = np.arange(10)
plt.errorbar(x, mean_1, yerr=std_1, label='Data 1', fmt='-o')
plt.errorbar(x, mean_2, yerr=std_2, label='Data 2', fmt='-s')
plt.yscale('log')
plt.show()
```
